```python
import json

input_file_path = '/mnt/petrelfs/share_data/yanziang/internvl/data/allseeing.jsonl'  # 原始 jsonl 文件
output_file_path = '/mnt/petrelfs/share_data/yanziang/internvl/data/allseeing_new.jsonl'  # 修改后的 jsonl 文件

# 读取 jsonl 文件并修改字典
with open(input_file_path, 'r', encoding='utf-8') as infile, open(output_file_path, 'w', encoding='utf-8') as outfile:
    for line in infile:
        # 将每行 JSON 字符串解析为字典
        data = json.loads(line)
        
        # 检查 'test' 字典并修改其值
        if 'name' in data:
            data['name'] = data['name'].replace('/mnt/petrelfs/share_data/yanziang/internvl/data/allseeing_feature/','sssdza:s3://yanziang/allseeing_sam_feature/')
        
        # 将修改后的字典转换为 JSON 字符串并写入新文件
        json.dump(data, outfile)
        outfile.write('\n')  # 添加换行符以分隔每个 JSON 对象
```
