```python
import json
import os
from tqdm import tqdm
import cv2
import numpy as np
from tqdm import tqdm
from petrel_client.client import Client
c = Client()
def save_video(path, video_path):
    images = [img for img in os.listdir(path) if img.endswith(".jpg")]
    images.sort()  

    first_image = cv2.imread(os.path.join(path, images[0]))
    height, width, layers = first_image.shape

    # video_name = 'output_video.mp4'
    # fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # 使用mp4编码
    # video = cv2.VideoWriter(video_name, fourcc, 20, (width, height))  # 1帧每秒

    # # 遍历所有图片并写入视频
    # for image in images:
    #     img_path = os.path.join(path, image)
    #     video.write(cv2.imread(img_path))
    # video.release()
    # cv2.destroyAllWindows()
    # with open('output_video.mp4', 'rb') as file_data:
    #     c.put(video_path, file_data)
    return [height, width]

fin = []
for dd in tqdm(os.listdir('/mnt/petrelfs/share/videointern/GOT/OpenDataLab___GOT-10k/raw/GOT-10k/val')):
    if '.txt' not in dd:
        path = os.path.join('/mnt/petrelfs/share/videointern/GOT/OpenDataLab___GOT-10k/raw/GOT-10k/val', dd)
        txt_file = os.path.join(path, 'groundtruth.txt')
        video_path = 'shdd:s3://yanziang/got10k/' + dd + '.mp4'
        with open(txt_file, 'r') as file:
            first_line = file.readline()  # 读取第一行

        # 将读取的字符串分割并转换为浮点数数组
        float_array = list(map(float, first_line.strip().split(',')))
        size_hw = save_video(path, video_path)
        float_array = [float_array[0], float_array[1], float_array[0] + float_array[2], float_array[1] + float_array[3]]
        item = {'video_path': video_path, 'bbox':float_array, 'size_hw':size_hw}
        # print(item)
        # break
        fin.append(item)
with open('got10k.json', 'w') as f:
    json.dump(fin, f)
```
